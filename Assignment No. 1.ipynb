# 7

```python
print("Enter Marks of Different Subjects (MAX 100)")
a = int(input("Urdu Marks: "))
b = int(input("English Marks: "))
c = int(input("Sindhi Marks: "))
d = int(input("Biology Marks: "))
e = int(input("Phyics Marks: "))

total= a+b+c+d+e
print(f"Your Total Marks are {total} out of 500")

percentage = round(total/500*100,1)

print(f"Your percentage is {percentage}%")
print("Your Grade is ")

if percentage <= 49:
    print("E")
elif percentage <= 59:
    print("C")
elif percentage <= 69:
    print("B")
elif percentage <= 79:
    print("A")
elif percentage <= 100:
    print("A+")
```
